```python
model = YOLO('yolov8n.pt')
camera_source = 1  # yoki RTSP manzili

cap = cv2.VideoCapture(camera_source)
results_list = []

frame_count = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Obyekt deteksiyasi
    results = model(frame)
    annotated_frame = results[0].plot()

    # Detected objects (masalan, rulon va defektlar)
    for box in results[0].boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        # Bu yerda 'cls' va 'conf' ni o'zingizga moslab ishlating
        results_list.append({
            'frame': frame_count,
            'class': cls,
            'confidence': conf,
            'x1': x1, 'y1': y1, 'x2': x2, 'y2': y2
        })
        # Defekt joyini chizish
        cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0,0,255), 2)

    cv2.imshow("Detection", annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

    frame_count += 1

cap.release()
cv2.destroyAllWindows()

# Natijalarni DataFrame va CSV ko‘rinishida saqlash
df = pd.DataFrame(results_list)
df.to_csv("detection_results.csv", index=False)
```
