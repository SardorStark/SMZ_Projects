# Connect to a Server from Jupyter Notebook

This notebook demonstrates how to connect to a server from Jupyter using Python. You will learn how to install required libraries, set connection parameters, establish a connection, and send/receive data.

## 1. Install Required Libraries

We will use `requests` for HTTP connections and `paramiko` for SSH connections. Run the following cell to install them if needed.

## 2. Set Server Connection Parameters

Define the server address, port, and credentials for both HTTP and SSH connections.

```python
# HTTP server parameters
http_url = "https://jsonplaceholder.typicode.com/posts/1"

# SSH server parameters
ssh_host = "your.ssh.server.com"
ssh_port = 22
ssh_username = "your_username"
ssh_password = "your_password"
```

## 3. Establish Server Connection

Connect to the HTTP server using `requests` and to the SSH server using `paramiko`.

```python
import requests
import paramiko

# Connect to HTTP server
http_response = requests.get(http_url)

# Connect to SSH server
ssh_client = paramiko.SSHClient()
ssh_client.set_missing_host_key_policy(paramiko.AutoAddPolicy())
try:
    ssh_client.connect(
        hostname=ssh_host,
        port=ssh_port,
        username=ssh_username,
        password=ssh_password
    )
    ssh_connected = True
except Exception as e:
    print(f"SSH connection failed: {e}")
    ssh_connected = False
```

## 4. Send and Receive Data

Send a request to the HTTP server and execute a command on the SSH server. Display the results.

```python
# Handle HTTP response
if http_response.status_code == 200:
    print("HTTP Response Data:")
    print(http_response.json())
else:
    print(f"HTTP request failed with status code {http_response.status_code}")

# Execute command on SSH server
if ssh_connected:
    stdin, stdout, stderr = ssh_client.exec_command("uname -a")
    print("SSH Command Output:")
    print(stdout.read().decode())
    ssh_client.close()
else:
    print("SSH connection not established.")
```
